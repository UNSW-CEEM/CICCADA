```python
import matplotlib.pyplot as plt

test_site = df[df["site_alias"] == "AUS001"].copy()
test_site["timestamp"] = pd.to_datetime(test_site["timestamp"] )
test_site = test_site.sort_values("timestamp")

start = test_site["timestamp"].min()
end = start + pd.Timedelta(days=7)
week_site = test_site[(test_site["timestamp"] >= start) & (test_site["timestamp"] < end)].copy()

week_site["active_power_total"] = week_site[["active_power_1", "active_power_2", "active_power_3"]].sum(axis=1)
week_site["reactive_power_total"] = week_site[["reactive_power_1", "reactive_power_2", "reactive_power_3"]].sum(axis=1)
week_site["voltage_avg"] = week_site[["ac_voltage_1", "ac_voltage_2", "ac_voltage_3"]].mean(axis=1)

fig, axes = plt.subplots(3, 1, sharex=True, figsize=(14, 11))
week_site.plot(x="timestamp", y="active_power_total", ax=axes[0], legend=False, color="tab:blue")
axes[0].set_title(f"Active power over 1 week for AUS001 ({start:%Y-%m-%d} to {end:%Y-%m-%d})")
axes[0].set_ylabel("Active power")
axes[0].grid(True, alpha=0.3)

week_site.plot(x="timestamp", y="reactive_power_total", ax=axes[1], legend=False, color="tab:orange")
axes[1].set_title("Reactive power over the same 1-week window")
axes[1].set_ylabel("Reactive power")
axes[1].grid(True, alpha=0.3)

week_site.plot(x="timestamp", y="voltage_avg", ax=axes[2], legend=False, color="tab:green")
axes[2].set_title("Average AC voltage over the same 1-week window")
axes[2].set_ylabel("Voltage")
axes[2].set_xlabel("Timestamp")
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
```
