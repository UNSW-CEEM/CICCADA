```python
site_id = 636269142
year, month = 2024, 3
df = iceberg_sql(f"""with data as (select t_stamp, circuit_type, sum(power*circuit_polarity/1000) as P_kw, 
                 sum(energy_reactive*circuit_polarity/1000*12) as Q_kvar, n_lat, n_long
                 from ts inner join meta_up23c as m on ts.circuit_id = m.circuit_id
                 where site_id = {site_id} and year = {year} and month = {month} and (circuit_type like 'ac_%' or circuit_type like 'pv_%')
                 group by t_stamp, circuit_type, n_lat, n_long),

                 bom10min as (select time, b.latitude, b.longitude, surface_global_irradiance as GHI, cloud_type,
                 lead(time) OVER (ORDER BY time) AS next_time,
                 lead(surface_global_irradiance)  OVER (ORDER BY time) AS next_GHI,
                 lead(cloud_type)  OVER (ORDER BY time) AS next_cloud_type
                 from bom_nci.solar as b
                 inner join meta_up23c as m on b.latitude = m.n_lat and b.longitude = m.n_long
                 where year={year} and month={month} and site_id = {site_id}),

                  bom5min as ((select time as time_5min, latitude, longitude, GHI, cloud_type
                 from bom10min) union all
                 (select date_add('minute', 5, time) as time_5min, latitude, longitude,
                 GHI + 0.5 * (next_GHI - GHI) AS GHI,
                 cloud_type + 0.5 * (next_cloud_type - cloud_type) AS cloud_type
                 FROM bom10min
                  WHERE next_time IS NOT NULL
                  ORDER BY time_5min)),
                  clear_sky as (select date_trunc('day', time + interval '10' hour) AS day
                  from bom10min
                  group by date_trunc('day', time + interval '10' hour)
                  having sum(cloud_type) > -1
                  )

                  select  d.*, GHI, cloud_type
                 from data as d left join bom5min as b on 
                 d.n_lat = b.latitude and d.n_long = b.longitude and d.t_stamp = b.time_5min
                 where date_trunc('day', t_stamp + interval '10' hour)  in (select day from clear_sky)
               
""")
df["t_stamp"] = (
    pd.to_datetime(df["t_stamp"])
    .dt.tz_localize("utc")
    .dt.tz_convert(pytz.FixedOffset(10 * 60))
)
```
